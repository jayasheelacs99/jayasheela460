```python
N = 4

solution = [1, 3, 0, 2]

def heuristic(state):
    conflicts = 0

    for i in range(N):
        for j in range(i + 1, N):
            if state[i] == state[j]:
                conflicts += 1

            if abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def display_board(state):
    print("\nFinal Chess Board:\n")
    print("  0 1 2 3")
    print("  --------")

    for row in range(N):
        print(row, end="| ")

        for col in range(N):
            if state[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")

        print()


# Main program
print("Final Solution:", solution)
print("Heuristic Value:", heuristic(solution))

display_board(solution)
```
