```python
import math

board = [" "] * 9

def print_board():
    print()
    for i in range(0, 9, 3):
        print(f" {board[i]} | {board[i+1]} | {board[i+2]} ")
        if i < 6:
            print("---+---+---")
    print()

def winner(player):
    wins = [
        (0,1,2), (3,4,5), (6,7,8),
        (0,3,6), (1,4,7), (2,5,8),
        (0,4,8), (2,4,6)
    ]
    return any(all(board[i] == player for i in combo) for combo in wins)

def minimax(is_ai):
    if winner("X"):
        return 1
    if winner("O"):
        return -1
    if " " not in board:
        return 0

    if is_ai:
        best = -math.inf
        for i in range(9):
            if board[i] == " ":
                board[i] = "X"
                best = max(best, minimax(False))
                board[i] = " "
        return best
    else:
        best = math.inf
        for i in range(9):
            if board[i] == " ":
                board[i] = "O"
                best = min(best, minimax(True))
                board[i] = " "
        return best

def ai_move():
    best_score = -math.inf
    move = None

    for i in range(9):
        if board[i] == " ":
            board[i] = "X"
            score = minimax(False)
            board[i] = " "

            if score > best_score:
                best_score = score
                move = i

    board[move] = "X"

print("Tic-Tac-Toe")
print("AI = X, Human = O")
print("Positions are 1 to 9.")

while True:
    print_board()

    # Human move
    while True:
        try:
            pos = int(input("Your move (1-9): ")) - 1
            if 0 <= pos < 9 and board[pos] == " ":
                board[pos] = "O"
                break
            print("Invalid move. Try again.")
        except ValueError:
            print("Enter a number from 1 to 9.")

    if winner("O"):
        print_board()
        print("You win!")
        break

    if " " not in board:
        print_board()
        print("Draw!")
        break

    # AI move
    ai_move()
    print("AI played:")
   
    if winner("X"):
        print_board()
        print("AI wins!")
        break

    if " " not in board:
        print_board()
        print("Draw!")
        break
```

```python
# Vacuum Cleaner Problem
# A = Left Room, B = Right Room
# 1 = Dirty, 0 = Clean

room = {
    "A": 1,
    "B": 1
}

position = "A"

print("Initial State:")
print("Room A:", "Dirty" if room["A"] else "Clean")
print("Room B:", "Dirty" if room["B"] else "Clean")
print("Vacuum position:", position)

while room["A"] == 1 or room["B"] == 1:

    if room[position] == 1:
        print("\nVacuum is in Room", position)
        print("Action: SUCK")
        room[position] = 0

    else:
        if position == "A":
            position = "B"
            print("\nAction: MOVE RIGHT")
        else:
            position = "A"
            print("\nAction: MOVE LEFT")

print("\nFinal State:")
print("Room A: Clean")
print("Room B: Clean")
print("Vacuum position:", position)
print("All rooms are clean!")
```
