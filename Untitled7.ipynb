* A star using misplace tiles

```python
def h(state, goal):
    distance = 0

    for i in range(3):
        for j in range(3):
            tile = state[i][j]

            if tile != 0:
                for x in range(3):
                    for y in range(3):
                        if goal[x][y] == tile:
                            distance += abs(i - x) + abs(j - y)

    return distance


def A_star(start, goal):
    OPEN = [(h(start, goal), 0, start, [])]
    visited = set()

    while OPEN:
        OPEN.sort(key=lambda x: x[0])
        f, g, state, path = OPEN.pop(0)

        if state == goal:
            print("Goal reached!")
            for s in path + [state]:
                for row in s:
                    print(row)
                print()
            return

        key = tuple(map(tuple, state))
        if key in visited:
            continue
        visited.add(key)

        x, y = next(
            (i, j)
            for i in range(3)
            for j in range(3)
            if state[i][j] == 0
        )

        for dx, dy in [(0, 1), (0, -1), (1, 0), (-1, 0)]:
            nx, ny = x + dx, y + dy

            if 0 <= nx < 3 and 0 <= ny < 3:
                new_state = [row[:] for row in state]
                new_state[x][y], new_state[nx][ny] = (
                    new_state[nx][ny], new_state[x][y]
                )

                key2 = tuple(map(tuple, new_state))
                if key2 not in visited:
                    new_g = g + 1
                    new_f = new_g + h(new_state, goal)
                    OPEN.append(
                        (new_f, new_g, new_state, path + [state])
                    )

    print("No solution found")


start = [
    [1, 2, 3],
    [4, 0, 6],
    [7, 5, 8]
]

goal = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0]
]

A_star(start, goal)
```
