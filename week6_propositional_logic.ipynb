```python

from itertools import product

# Implication function
def implies(a, b):
    return not a or b

# Knowledge Base
def KB(P, Q, R):
    return (implies(Q, P) and
            implies(P, not Q) and
            (Q or R))

# Truth table
print("TRUTH TABLE")
print("P Q R | Q->P | P->~Q | Q|R | KB")
print("--------------------------------------")

for P, Q, R in product([True, False], repeat=3):

    a = implies(Q, P)
    b = implies(P, not Q)
    c = Q or R
    kb = a and b and c

    print(int(P), int(Q), int(R), "|",
          int(a), "   |", int(b), "    |",
          int(c), "  |", int(kb))

# Entailment checking
def entails(query):
    for P, Q, R in product([True, False], repeat=3):
        if KB(P, Q, R) and not query(P, Q, R):
            return False
    return True

print("\nENTAILMENT RESULTS")

print("KB entails R:",
      entails(lambda P, Q, R: R))

print("KB entails R -> P:",
      entails(lambda P, Q, R: implies(R, P)))

print("KB entails Q -> R:",
      entails(lambda P, Q, R: implies(Q, R)))
```
