```python

# First Order Logic - Forward Chaining

# Initial facts
facts = [
    ("American", "Robert"),
    ("Enemy", "A", "America"),
    ("Owns", "A", "T1"),
    ("Missile", "T1")
]

# Forward Chaining
def forward_chaining(facts):

    # Rule 1: Missiles are weapons
    for f in facts:
        if f[0] == "Missile":
            new_fact = ("Weapon", f[1])
            if new_fact not in facts:
                facts.append(new_fact)
                print("Derived:", new_fact)

    # Rule 2: Robert sells missiles owned by A to A
    for f in facts:
        if f[0] == "Missile":
            missile = f[1]
            if ("Owns", "A", missile) in facts:
                new_fact = ("Sells", "Robert", missile, "A")
                if new_fact not in facts:
                    facts.append(new_fact)
                    print("Derived:", new_fact)

    # Rule 3: Enemy of America is hostile
    for f in facts:
        if f[0] == "Enemy":
            new_fact = ("Hostile", f[1])
            if new_fact not in facts:
                facts.append(new_fact)
                print("Derived:", new_fact)

    # Rule 4: American selling weapons to hostile nations is criminal
    for f in facts:
        if f[0] == "Sells":
            person, weapon, country = f[1], f[2], f[3]

            if (("American", person) in facts and
                ("Weapon", weapon) in facts and
                ("Hostile", country) in facts):

                new_fact = ("Criminal", person)
                if new_fact not in facts:
                    facts.append(new_fact)
                    print("Derived:", new_fact)

    return facts


# Execute
result = forward_chaining(facts)

# Final result
if ("Criminal", "Robert") in result:
    print("\nRobert is Criminal")
else:
    print("\nRobert is Not Criminal")
```
