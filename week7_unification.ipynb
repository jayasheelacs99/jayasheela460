```python

def unify(a, b, subst):

    if a == b:
        return subst

    if isinstance(a, str) and a.islower():
        if a in subst:
            return unify(subst[a], b, subst)

        subst[a] = b
        return subst

    if isinstance(b, str) and b.islower():
        if b in subst:
            return unify(a, subst[b], subst)

        subst[b] = a
        return subst

    if isinstance(a, tuple) and isinstance(b, tuple):

        if a[0] != b[0] or len(a) != len(b):
            return None

        for x, y in zip(a[1:], b[1:]):
            subst = unify(x, y, subst)

            if subst is None:
                return None

        return subst

    return None


# Example 1: Successful
a = ("Likes", "Rahul", "Mango")
b = ("Likes", "x", "Mango")

print("Example 1: Successful")
result = unify(a, b, {})

if result is not None:
    print("Unification Successful")
    print("Substitution:", result)
else:
    print("Unification Failed")


# Example 2: Failed
a = ("Knows", "John", "x")
b = ("Knows", "x", "Elisabeth")

print("\nExample 2: Failed")
result = unify(a, b, {})

if result is not None:
    print("Unification Successful")
    print("Substitution:", result)
else:
    print("Unification Failed")
```
